```python
import os
import cv2
import glob
import numpy as np
import random
import matplotlib.pyplot as plt
from skimage.transform import resize
from sklearn.model_selection import train_test_split
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping, ReduceLROnPlateau, Callback
import pandas as pd
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# Set random seeds for reproducibility
random.seed(42)
np.random.seed(42)
tf.random.set_seed(42)

# Image dimensions
IMG_HEIGHT, IMG_WIDTH, IMG_CHANNELS = 352, 352, 3

# Dataset paths (adjust if different)
TRAIN_IMAGE_PATH = 'Train-Input1'  # jpg
TRAIN_MASK_PATH = 'Train-Mask1'    # jpg
TEST_IMAGE_PATH = 'Test-Input'    # jpg
TEST_MASK_PATH = 'Test-Mask'      # jpg

# Mount Google Drive


# Custom metrics
def specificity(y_true, y_pred):
    true_negatives = tf.reduce_sum(tf.round(tf.clip_by_value((1 - y_true) * (1 - y_pred), 0, 1)))
    possible_negatives = tf.reduce_sum(tf.round(tf.clip_by_value(1 - y_true, 0, 1)))
    return true_negatives / (possible_negatives + tf.keras.backend.epsilon())

def dice_score(y_true, y_pred):
    y_true_f = tf.keras.backend.flatten(y_true)
    y_pred_f = tf.keras.backend.flatten(y_pred)
    intersection = tf.keras.backend.sum(y_true_f * y_pred_f)
    return (2. * intersection + tf.keras.backend.epsilon()) / (
        tf.keras.backend.sum(y_true_f) + tf.keras.backend.sum(y_pred_f) + tf.keras.backend.epsilon()
    )

def iou_score(y_true, y_pred):
    y_true_f = tf.keras.backend.flatten(y_true)
    y_pred_f = tf.keras.backend.flatten(tf.clip_by_value(y_pred, 0, 1))
    intersection = tf.keras.backend.sum(y_true_f * y_pred_f)
    union = tf.keras.backend.sum(y_true_f) + tf.keras.backend.sum(y_pred_f) - intersection
    return (intersection + tf.keras.backend.epsilon()) / (union + tf.keras.backend.epsilon())

# Custom loss with higher positive class weight
def weighted_binary_focal_dice_loss(y_true, y_pred):
    alpha = 0.75
    gamma = 2.0
    bce = tf.keras.losses.binary_crossentropy(y_true, y_pred, from_logits=False)
    bce_exp = tf.exp(-bce)
    focal_loss = alpha * (1 - bce_exp) ** gamma * bce
    focal_loss = tf.reduce_mean(focal_loss)
    
    intersection = tf.reduce_sum(y_true * y_pred)
    dice_loss = 1 - (2. * intersection + tf.keras.backend.epsilon()) / (
        tf.reduce_sum(y_true) + tf.reduce_sum(y_pred) + tf.keras.backend.epsilon()
    )
    
    weights = tf.where(y_true > 0, 50.0, 1.0)  # Increased to 50 for positive class
    weighted_focal = tf.reduce_mean(focal_loss * weights)
    
    return weighted_focal + dice_loss

# Custom callback with lower threshold
class MetricsLogger(Callback):
    def __init__(self, validation_data, threshold=0.2):
        super().__init__()
        self.validation_data = validation_data
        self.threshold = threshold
        self.metrics_log = []

    def on_epoch_end(self, epoch, logs=None):
        logs = logs or {}
        val_x, val_y = self.validation_data
        val_pred = self.model.predict(val_x, verbose=0)
        val_pred_t = (val_pred > self.threshold).astype(np.float32)
        
        val_loss = logs.get('val_loss', 0)
        val_accuracy = logs.get('val_accuracy', 0)
        val_dice = dice_score(val_y, val_pred_t)
        val_iou = iou_score(val_y, val_pred_t)
        val_precision = tf.keras.metrics.Precision()(val_y, val_pred_t)
        val_recall = tf.keras.metrics.Recall()(val_y, val_pred_t)
        val_specificity = specificity(val_y, val_pred_t)
        
        log_entry = {
            'epoch': epoch + 1,
            'val_loss': val_loss,
            'val_accuracy': val_accuracy,
            'val_dice_score': val_dice.numpy(),
            'val_iou_score': val_iou.numpy(),
            'val_precision': val_precision.numpy(),
            'val_recall': val_recall.numpy(),
            'val_specificity': val_specificity.numpy()
        }
        self.metrics_log.append(log_entry)
        
        print(f"\n--- Epoch {epoch + 1} Validation Metrics (threshold={self.threshold}) ---")
        for key, value in log_entry.items():
            if key != 'epoch':
                print(f"{key}: {value:.4f}")
        
        if (epoch + 1) % 5 == 0:
            df = pd.DataFrame(self.metrics_log)
            df.to_csv('metrics_log.csv', index=False)
            print(f"Metrics up to epoch {epoch + 1} saved to metrics_log.csv")

    def on_train_end(self, logs=None):
        df = pd.DataFrame(self.metrics_log)
        df.to_csv('unet_metrics_log.csv', index=False)
        print("Full metrics log saved to Google Drive!")

# Lightweight U-Net with dropout
def build_unet(input_shape=(IMG_HEIGHT, IMG_WIDTH, IMG_CHANNELS)):
    inputs = layers.Input(input_shape)
    
    # Encoder
    c1 = layers.Conv2D(32, 3, padding='same', activation='relu')(inputs)
    c1 = layers.Conv2D(32, 3, padding='same', activation='relu')(c1)
    p1 = layers.MaxPooling2D(2)(c1)
    c2 = layers.Conv2D(64, 3, padding='same', activation='relu')(p1)
    c2 = layers.Conv2D(64, 3, padding='same', activation='relu')(c2)
    p2 = layers.MaxPooling2D(2)(c2)
    c3 = layers.Conv2D(128, 3, padding='same', activation='relu')(p2)
    c3 = layers.Conv2D(128, 3, padding='same', activation='relu')(c3)
    p3 = layers.MaxPooling2D(2)(c3)
    
    # Bottleneck
    b = layers.Conv2D(256, 3, padding='same', activation='relu')(p3)
    b = layers.Conv2D(256, 3, padding='same', activation='relu')(b)
    b = layers.Dropout(0.3)(b)  # Added dropout
    
    # Decoder
    u4 = layers.UpSampling2D(2)(b)
    u4 = layers.Concatenate()([u4, c3])
    c4 = layers.Conv2D(128, 3, padding='same', activation='relu')(u4)
    c4 = layers.Conv2D(128, 3, padding='same', activation='relu')(c4)
    c4 = layers.Dropout(0.2)(c4)
    u5 = layers.UpSampling2D(2)(c4)
    u5 = layers.Concatenate()([u5, c2])
    c5 = layers.Conv2D(64, 3, padding='same', activation='relu')(u5)
    c5 = layers.Conv2D(64, 3, padding='same', activation='relu')(c5)
    u6 = layers.UpSampling2D(2)(c5)
    u6 = layers.Concatenate()([u6, c1])
    c6 = layers.Conv2D(32, 3, padding='same', activation='relu')(u6)
    c6 = layers.Conv2D(32, 3, padding='same', activation='relu')(c6)
    
    outputs = layers.Conv2D(1, 1, activation='sigmoid')(c6)
    return models.Model(inputs, outputs)

# Data loading with debug
def load_data(image_path, mask_path, target_size=(IMG_HEIGHT, IMG_WIDTH)):
    image_files = sorted(glob.glob(os.path.join(image_path, '*.jpg')))
    mask_files = sorted(glob.glob(os.path.join(mask_path, '*.png')))
    
    min_len = min(len(image_files), len(mask_files))
    image_files = image_files[:min_len]
    mask_files = mask_files[:min_len]
    
    images = np.zeros((min_len, *target_size, IMG_CHANNELS), dtype=np.float32)
    masks = np.zeros((min_len, *target_size, 1), dtype=np.float32)
    
    for i, (img_file, mask_file) in enumerate(zip(image_files, mask_files)):
        img = cv2.imread(img_file)
        if img is None:
            print(f"Warning: Could not load image {img_file}")
            continue
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        img = resize(img, target_size, mode='constant', preserve_range=True) / 255.0
        images[i] = img
        
        mask = cv2.imread(mask_file, cv2.IMREAD_GRAYSCALE)
        if mask is None:
            print(f"Warning: Could not load mask {mask_file}")
            continue
        mask = resize(mask, target_size, mode='constant', preserve_range=True)
        mask = np.expand_dims((mask > 0).astype(np.float32), -1)  # Threshold >0
        masks[i] = mask
    
    positive_pixels = np.sum(masks > 0) / (min_len * target_size[0] * target_size[1])
    print(f"Loaded {len(images)} paired samples from {image_path}")
    print(f"Positive pixels: {positive_pixels:.4f}%")
    return images, masks

# Load data
print("Loading training data...")
Train_Inputs, Train_Masks = load_data(TRAIN_IMAGE_PATH, TRAIN_MASK_PATH)
print("Loading test data...")
Test_Inputs, Test_Masks = load_data(TEST_IMAGE_PATH, TEST_MASK_PATH)

# Check validation split
val_size = 100 / len(Train_Inputs)
Train_Inputs, Val_Inputs, Train_Masks, Val_Masks = train_test_split(
    Train_Inputs, Train_Masks, test_size=val_size, random_state=42
)
print(f"Train: {len(Train_Inputs)}, Val: {len(Val_Inputs)}, Test: {len(Test_Inputs)}")
val_positive_pixels = np.sum(Val_Masks > 0) / (len(Val_Inputs) * IMG_HEIGHT * IMG_WIDTH)
print(f"Validation positive pixels: {val_positive_pixels:.4f}%")
if val_positive_pixels < 0.001:
    print("Warning: Validation set has very few positive pixels. Consider reshuffling.")

# Data augmentation
data_gen_args = dict(
    rotation_range=45,      # Increased
    width_shift_range=0.3,
    height_shift_range=0.3,
    shear_range=0.2,        # Added shear
    zoom_range=0.4,
    horizontal_flip=True,
    vertical_flip=True,
    fill_mode='nearest'
)
image_datagen = ImageDataGenerator(**data_gen_args)
mask_datagen = ImageDataGenerator(**data_gen_args)

def data_generator(images, masks, batch_size):
    seed = 42
    image_datagen.fit(images, augment=True, seed=seed)
    mask_datagen.fit(masks, augment=True, seed=seed)
    image_gen = image_datagen.flow(images, batch_size=batch_size, seed=seed)
    mask_gen = mask_datagen.flow(masks, batch_size=batch_size, seed=seed)
    for img_batch, mask_batch in zip(image_gen, mask_gen):
        yield img_batch, mask_batch

# Create and compile model
model = build_unet()
model.compile(
    optimizer=Adam(learning_rate=0.0001),
    loss=weighted_binary_focal_dice_loss,
    metrics=['accuracy', dice_score, specificity, iou_score, 'precision', 'recall']
)
print(model.summary())

# Callbacks with mode='max'
callbacks = [
    ModelCheckpoint('best_unet_model.h5', monitor='val_dice_score', save_best_only=True, mode='max', verbose=1),
    EarlyStopping(monitor='val_dice_score', patience=15, restore_best_weights=True, mode='max', verbose=1),
    ReduceLROnPlateau(monitor='val_dice_score', factor=0.5, patience=5, min_lr=1e-6, mode='max', verbose=1),
    MetricsLogger((Val_Inputs, Val_Masks), threshold=0.2)
]

# Train
print("Starting training...")
history = model.fit(
    data_generator(Train_Inputs, Train_Masks, batch_size=8),
    steps_per_epoch=len(Train_Inputs) // 8,
    epochs=30,  # Reduced for faster iteration
    validation_data=(Val_Inputs, Val_Masks),
    callbacks=callbacks,
    verbose=1
)

# Save model
model.save('unet_model_fixed.h5')
print("Model saved!")

# Plot history
fig, axes = plt.subplots(2, 3, figsize=(18, 8))
axes[0, 0].plot(history.history['loss'], label='Train Loss')
axes[0, 0].plot(history.history['val_loss'], label='Val Loss')
axes[0, 0].set_title('Loss')
axes[0, 0].legend()

axes[0, 1].plot(history.history['accuracy'], label='Train Acc')
axes[0, 1].plot(history.history['val_accuracy'], label='Val Acc')
axes[0, 1].set_title('Accuracy')
axes[0, 1].legend()

axes[0, 2].plot(history.history['dice_score'], label='Train Dice')
axes[0, 2].plot(history.history['val_dice_score'], label='Val Dice')
axes[0, 2].set_title('Dice Score')
axes[0, 2].legend()

axes[1, 0].plot(history.history['iou_score'], label='Train IoU')
axes[1, 0].plot(history.history['val_iou_score'], label='Val IoU')
axes[1, 0].set_title('IoU Score')
axes[1, 0].legend()

axes[1, 1].plot(history.history['precision'], label='Train Precision')
axes[1, 1].plot(history.history['val_precision'], label='Val Precision')
axes[1, 1].set_title('Precision')
axes[1, 1].legend()

axes[1, 2].plot(history.history['recall'], label='Train Recall')
axes[1, 2].plot(history.history['val_recall'], label='Val Recall')
axes[1, 2].set_title('Recall')
axes[1, 2].legend()

plt.tight_layout()
plt.show()

# Evaluate on test set
print("Evaluating on test set...")
test_results = model.evaluate(Test_Inputs, Test_Masks, verbose=1)
metric_names = ['loss', 'accuracy', 'dice_score', 'specificity', 'iou_score', 'precision', 'recall']
for name, value in zip(metric_names, test_results):
    print(f"Test {name}: {value:.4f}")

# Predict and visualize
preds_test = model.predict(Test_Inputs, verbose=1)
preds_test_t = (preds_test > 0.2).astype(np.uint8)

# Show two random test samples
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
for i in range(2):
    ix = random.randint(0, len(Test_Masks) - 1)
    axes[i, 0].imshow(Test_Inputs[ix])
    axes[i, 0].set_title('Test Image')
    axes[i, 1].imshow(np.squeeze(preds_test_t[ix]), cmap='gray')
    axes[i, 1].set_title('Predicted Plaque Mask')
    axes[i, 2].imshow(np.squeeze(Test_Masks[ix]), cmap='gray')
    axes[i, 2].set_title('Ground Truth Mask')
plt.tight_layout()
plt.show()

# Debug mask values
sample_mask = cv2.imread(glob.glob(os.path.join(TRAIN_MASK_PATH, '*.jpg'))[0], cv2.IMREAD_GRAYSCALE)
print(f"Sample mask unique values: {np.unique(sample_mask)}")
print("Done! Metrics saved to unet_metrics_log.csv")
```
